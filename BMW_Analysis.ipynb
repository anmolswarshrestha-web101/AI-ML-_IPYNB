```python
#-----------------------------------------------------------------------------
# 1. DATA ANALYSIS CALCULATIONS
#-----------------------------------------------------------------------------

total_records = len(df)
avg_price = df['Price_USD'].mean()
total_sales_volume = df['Sales_Volume'].sum()
top_model = df['Model'].value_counts().idxmax()
best_region = df.groupby('Region')['Sales_Volume'].sum().idxmax()
electric_percentage = (len(df[df['Fuel_Type'] == 'Electric']) / total_records) * 100
ct_region_color = pd.crosstab(df['Region'], df['Color'])

# Transmission & Region-Color calculations
trans_median_price = df.groupby('Transmission')['Price_USD'].median()
top_transmission = trans_median_price.idxmax()
top_transmission_price = trans_median_price.max()

stacked_ct = ct_region_color.stack()
top_region_color_combo = stacked_ct.idxmax()
top_region_color_count = stacked_ct.max()

#------------------------------------------------------------------------------
# 2. THE VISUAL REPORT (4-Chart Dashboard)
#------------------------------------------------------------------------------
fig, axes = plt.subplots(3, 2, figsize=(16, 18))
fig.suptitle('BMW MARKET ANALYSIS EXECUTIVE REPORT (2010-2024)', fontsize=20, fontweight='bold', y=0.95)

#------------------------------------------------------------------------------
yearly_sales = df.groupby('Year')['Sales_Volume'].sum()
axes[0, 0].plot(yearly_sales.index, yearly_sales.values, marker='s', linewidth=2)
axes[0, 0].set_title('Annual Sales Volume Trend', fontsize=14)
axes[0, 0].set_ylabel('Units Sold')
axes[0, 0].grid(True, alpha=0.3)

#--- CHART 2: Regional Market Share ---
region_sales = df.groupby('Region')['Sales_Volume'].sum().sort_values()
axes[0, 1].barh(region_sales.index, region_sales.values, color='teal')
axes[0, 1].set_title('Sales Volume By Region', fontsize=14)
axes[0, 1].set_xlabel('Total Units')

#--- CHART 3: Fuel Type Distribution ---
fuel_counts = df['Fuel_Type'].value_counts()
axes[1, 0].pie(fuel_counts, labels=fuel_counts.index, autopct='%1.1f%%')
axes[1, 0].set_title('Fuel Type Inventory Share', fontsize=14)

#--- CHART 4: Price vs Sales Volume (Sampled) ---
# We sample 1000 points so it doesn't look like a solid block
sample_df = df.sample(1000)
axes[1, 1].scatter(sample_df['Price_USD'], sample_df['Sales_Volume'], alpha=0.5, color='purple')
axes[1, 1].set_title('Price vs Sales Volume (Correlation)', fontsize=14)
axes[1, 1].set_xlabel('Price (USD)')
axes[1, 1].set_ylabel('Sales Volume')

# --- CHART 5: Boxplot of Price by Transmission ---
df.boxplot(column='Price_USD', by='Transmission', grid=False, ax=axes[2, 0])
axes[2, 0].set_title('Price Distribution by Transmission Type', fontsize=14)
axes[2, 0].set_ylabel('Price (USD)')
axes[2, 0].set_xlabel('Transmission')

# --- CHART 6: Crosstab Heatmap of Region vs Color ---
sns.heatmap(ct_region_color, annot=True, fmt="d", cmap="YlGnBu", ax=axes[2, 1])
axes[2, 1].set_title('Region vs Color (Counts)', fontsize=14)
axes[2, 1].set_ylabel('Region')
axes[2, 1].set_xlabel('Color')

# Set the report title LAST, after all plotting, so it's not overwritten
plt.suptitle('')  # clears pandas' auto title for boxplot
fig.suptitle('BMW MARKET ANALYSIS EXECUTIVE REPORT (2010-2024)', fontsize=20, fontweight='bold', y=0.95)



plt.tight_layout(rect=[0, 0.03, 1, 0.92]) #left =0, top =0.92, right=1, bottom=0.03
plt.show()

# ==================================================================
# 3. TEXT-BASED EXECUTIVE SUMMARY DOC
# ==================================================================

print("="*60)
print("            BMW DATA SCIENCE ANALYSIS REPORT")
print("="*60)
print(f"REPORT SCOPE:        2010 - 2024")
print(f"TOTAL RECORDS:       {total_records:,} Units Analyzed")
print("-" * 60)
print(f"1. FINANCIALS:       Average Sale Price is ${avg_price:,.2f}")
print(f"2. TOP PERFORMER:    The '{top_model}' is the most frequent model in stock.")
print(f"3. MARKET LEADER:    '{best_region}' is the highest performing sales region.")
print(f"4. EV ADOPTION:      Electric vehicles currently make up {electric_percentage:.1f}% of inventory.")
print(f"5. TRANSMISSION:     '{top_transmission}' vehicles carry the higher median")
print(f"                     price, at ${top_transmission_price:,.2f}.")
print(f"6. TOP COMBO:        '{top_region_color_combo[1]}' is the most common color in")
print(f"                     '{top_region_color_combo[0]}', with {top_region_color_count:,} units.")
print("-" * 60)
print("REGION x COLOR CROSSTAB (Vehicle Counts):")
print(ct_region_color)
print("-" * 60)
print("CRITICAL INSIGHTS:")
print("- The sales volume shows a remarkably stable trend over the 14-year period.")
print("- Pricing is evenly distributed, suggesting BMW maintains a consistent")
print("  presence across both 'Entry-level Luxury' and 'High-end Luxury' tiers.")
print("- High overlap in the Scatter Plot indicates that Price does not")
print("  significantly restrict Sales Volume in this specific dataset.")
print("- Median pricing between Automatic and Manual transmissions is close,")
print("  indicating transmission type is not a strong price differentiator.")
print("- Color preference by region is fairly even, with no single color")
print("  dominating any one market - the strongest pairing is still only a")
print("  modest lead over the rest.")
print("="*60)
```

## Summary:

### 1) np.mean(list): Get the average.

### 2) np.sum(list): Add everything up.

### 3) np.max(list): Find the biggest number.

### 4) np.min(list): Find the smallest number.

### 5) round(number, 2): Makes the decimal shorts.

### 6) len(list): Counts how many items are in the list.


```python
# List of all our arrays to make the loop easy
columns = [year, engine, mileage, price, volume]
column_names = ["Year", "Engine_Size_(L)", "Mileage_(KM)", "Price_(USD)", "Sales_Volume"]
```

```python
import numpy as np

# 1. Group your data variables into a list called 'columns'
columns = [price, mileage, engine, volume]

# 2. Run your loop
summary_list = []

for data in columns:
    stats = {
        "Mean (Avg)": round(np.mean(data), 2),
        "Std Dev (Spread)": round(np.std(data), 2),
        "Total (Sum)": np.sum(data),
        "Highest (Max)": np.max(data),
        "Lowest (Min)": np.min(data),
        "Count": len(data)
    }
    summary_list.append(stats)

print(summary_list)
```

```python
import pandas as pd

# 1. Define row labels corresponding to your data columns
column_names = ["Price", "Mileage", "Engine", "Volume"]  # Adjust labels to match your summary_list

# 2. Create the table
summary_table = pd.DataFrame(summary_list, index=column_names)

# 3. Display the table
summary_table
```
