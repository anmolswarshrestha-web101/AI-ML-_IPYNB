```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# ==============================================================================
# 1. DATA ANALYSIS CALCULATIONS
# ==============================================================================

total_records = len(df)
avg_score = df["Average_Score"].mean()
total_hours_studied = df["Hours_Studied"].sum()
top_grade = df["Grade"].value_counts().idxmax()
best_result_group = df.groupby("Result")["Average_Score"].mean().idxmax()
pass_percentage = (len(df[df["Result"] == "Pass"]) / total_records) * 100
ct_result_grade = pd.crosstab(df["Result"], df["Grade"])

# Transmission & Region-Color calculations adapted to Grade & Result-Grade metrics
age_median_score = df.groupby("Age")["Average_Score"].median()
top_age = age_median_score.idxmax()
top_age_score = age_median_score.max()

stacked_ct = ct_result_grade.stack()
top_result_grade_combo = stacked_ct.idxmax()
top_result_grade_count = stacked_ct.max()

# ==============================================================================
# 2. THE VISUAL REPORT (6-Chart Dashboard)
# ==============================================================================

fig, axes = plt.subplots(3, 2, figsize=(16, 18))
fig.suptitle(
    "STUDENT PERFORMANCE ANALYSIS EXECUTIVE REPORT",
    fontsize=20,
    fontweight="bold",
    y=0.95,
)

# --- CHART 1: Average Score Trend by Age ---
age_scores = df.groupby("Age")["Average_Score"].mean()
axes[0, 0].plot(
    age_scores.index,
    age_scores.values,
    marker="s",
    linewidth=2,
    color="tab:blue",
)
axes[0, 0].set_title("Average Score Trend by Age", fontsize=14)
axes[0, 0].set_ylabel("Average Score")
axes[0, 0].grid(True, alpha=0.3)

# --- CHART 2: Total Hours Studied by Result ---
result_hours = df.groupby("Result")["Hours_Studied"].sum().sort_values()
axes[0, 1].barh(result_hours.index, result_hours.values, color="teal")
axes[0, 1].set_title("Total Hours Studied By Result", fontsize=14)
axes[0, 1].set_xlabel("Total Hours Studied")

# --- CHART 3: Grade Distribution ---
grade_counts = df["Grade"].value_counts()
axes[1, 0].pie(
    grade_counts, labels=grade_counts.index, autopct="%1.1f%%", startangle=90
)
axes[1, 0].set_title("Grade Share Distribution", fontsize=14)

# --- CHART 4: Hours Studied vs Average Score (Sampled) ---
# We sample 1000 points (or dataset size) so it doesn't look like a solid block
sample_df = df.sample(min(1000, len(df)))
axes[1, 1].scatter(
    sample_df["Hours_Studied"],
    sample_df["Average_Score"],
    alpha=0.5,
    color="purple",
)
axes[1, 1].set_title(
    "Hours Studied vs Average Score (Correlation)", fontsize=14
)
axes[1, 1].set_xlabel("Hours Studied")
axes[1, 1].set_ylabel("Average Score")

# --- CHART 5: Boxplot of Average Score by Age ---
df.boxplot(
    column="Average_Score", by="Age", grid=False, ax=axes[2, 0]
)
axes[2, 0].set_title("Score Distribution by Age Group", fontsize=14)
axes[2, 0].set_ylabel("Average Score")
axes[2, 0].set_xlabel("Age")

# --- CHART 6: Crosstab Heatmap of Result vs Grade ---
sns.heatmap(
    ct_result_grade,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    ax=axes[2, 1],
)
axes[2, 1].set_title("Result vs Grade (Counts)", fontsize=14)
axes[2, 1].set_ylabel("Result")
axes[2, 1].set_xlabel("Grade")

# Set the report title LAST, after all plotting, so it's not overwritten
plt.suptitle("")  # clears pandas' auto title for boxplot
fig.suptitle(
    "STUDENT PERFORMANCE ANALYSIS EXECUTIVE REPORT",
    fontsize=20,
    fontweight="bold",
    y=0.95,
)

plt.tight_layout(
    rect=[0, 0.03, 1, 0.92]
)  # left=0, top=0.92, right=1, bottom=0.03
plt.show()

# ==============================================================================
# 3. TEXT-BASED EXECUTIVE SUMMARY DOC
# ==============================================================================

print("=" * 60)
print("         STUDENT PERFORMANCE DATA SCIENCE ANALYSIS REPORT")
print("=" * 60)
print(f"REPORT SCOPE:         Academic Dataset Analysis")
print(f"TOTAL RECORDS:        {total_records:,} Students Analyzed")
print("-" * 60)
print(f"1. FINANCIALS:        Average Overall Score is {avg_score:.2f}")
print(f"2. TOP PERFORMER:     The '{top_grade}' grade is the most frequent grade assigned.")
print(f"3. MARKET LEADER:     '{best_result_group}' status represents the highest performing group.")
print(f"4. EV ADOPTION:       Passing students currently make up {pass_percentage:.1f}% of total cohort.")
print(f"5. TRANSMISSION:      Age group '{top_age}' students carry the higher median")
print(f"                      score, at {top_age_score:.2f}.")
print(f"6. TOP COMBO:         '{top_result_grade_combo[1]}' grade combined with '{top_result_grade_combo[0]}'")
print(f"                      status is the most common combination, with {top_result_grade_count:,} students.")
print("-" * 60)
print("RESULT x GRADE CROSSTAB (Student Counts):")
print(ct_result_grade)
print("-" * 60)
print("CRITICAL INSIGHTS:")
print("- Student average scores show a remarkably stable distribution across age groups.")
print("- Performance is evenly distributed, suggesting consistent academic standards across subjects.")
print("- High overlap in the Scatter Plot indicates study hours play a key role in score variance.")
print("- Median scores between different age categories remain close, showing age is not a sole differentiator.")
print("- Grade distribution across results is structured predictably with clear performance bands.")
print("=" * 60)
```

## Summary:

### np.mean(list): Get the average.

### np.sum(list): Add everything up.

### np.max(list): Find the biggest number.

### np.min(list): Find the smallest number.

### round(number, 2): Makes the decimal shorts.

### len(list): Counts how many items are in the list.

```python
# List of all our arrays to make the loop easy
columns = [age, math_score, science_score, english_score, hours_studied, total_score, average_score]
column_names = ["Age", "Math_Score", "Science_Score", "English_Score", "Hours_Studied", "Total_Score", "Average_Score"]
```

```python
import numpy as np

# 1. Group your data variables into a list called 'columns'
columns = [math_score, science_score, english_score, hours_studied, average_score]

# 2. Run your loop
summary_list = []

for data in columns:
    stats = {
        "Mean (Avg)": round(np.mean(data), 2),
        "Std Dev (Spread)": round(np.std(data), 2),
        "Total (Sum)": np.sum(data),
        "Highest (Max)": np.max(data),
        "Lowest (Min)": np.min(data),
        "Count": len(data)
    }
    summary_list.append(stats)

print(summary_list)
```

```python
import pandas as pd

# 1. Define row labels corresponding to your data columns
column_names = ["Math", "Science", "English", "Hours Studied", "Average Score"] # Adjust labels to match your summary_list

# 2. Create the table
summary_table = pd.DataFrame(summary_list, index=column_names)

# 3. Display the table
summary_table
```
