## STEP 8 - 2D Arrays (Like a Table!)

>A **2D Array** is like a table with rows and columns**
> Think of it like a school marks register!

```python
# Stack all three subject scores as columns gives us a 2D array
# Each row = one student, each column = one subject
scores_2d = np.column_stack([math, science, english])

print('Shape of 2D array:', scores_2d.shape)  # Should be (500, 3)
print('500 rows (students), 3 columns (subjects)\n')

print('First 5 rows (students):')
print('Math | Science | English')
print(scores_2d[:5])

print()

# Average score per SUBJECT (axis=0 → down the columns)
subject_avg = np.mean(scores_2d, axis=0)
print('Average by Subject:')
print(f' Math    : {subject_avg[0]:.2f}')
print(f' Science : {subject_avg[1]:.2f}')
print(f' English : {subject_avg[2]:.2f}')

print()

student_avg = np.mean(scores_2d, axis=1)
print('Average score per student (first 5):', np.round(student_avg[:5], 2))
# Average score per STUDENT (axis=1 B across the row)
```

```python
# np.unique( find all unique values (no repeats))
unique_ages = np.unique(ages)
print('Unique ages in our data:', unique_ages)

print()

# np.percentile() like "top X%"
p25 = np.percentile(math, 25)  # 25th percentile
p75 = np.percentile(math, 75)  # 75th percentile
print(f'Math Score Percentiles:')
print(f' 25th percentile: {p25}  -Bottom 25% scored below this')
print(f' 75th percentile: {p75}  - Top 25% scored above this')

print()

# np.clip()  limit values between a min and max
# Let's say max score allowed is 95
math_capped = np.clip(math, 0, 95)
print(f'Original max math score : {np.max(math)}')
print(f'After capping at 95     : {np.max(math_capped)}')

print()

#np.where() like IF-ELSE for arrays!
# Give grade 'Pass if score >=50, else 'Fail'
result = np.where(math >= 50, 'Pass', 'Fail')
pass_count = np.sum(result =='Pass')
fail_count = np.sum(result == 'Fail')
print(f'Math Pass/Fail:')
print(f'   Pass: {pass_count} students')
print(f'   Fail: {fail_count} students')

print()
result = np.where(science >= 50, 'Pass', 'Fail')
pass_count = np.sum(result =='Pass')
fail_count = np.sum(result == 'Fail')
print(f'science Pass/Fail:')
print(f'   Pass: {pass_count} students')
print(f'   Fail: {fail_count} students')

print()
```

```python
# np.corrcoef()   gives a correlation matrix
# We want: correlation between hours studied and math score

corr_matrix = np.corrcoef(hours, math)
correlation = corr_matrix[0, 1]   # top-right value in the 2x2 matrix

print(f'Correlation between Hours Studied & Math Score: {correlation:.4f}')
print()

if correlation > 0.5:
    print(' Strong positive link - studying more = better math!')
elif correlation > 0.2:
    print(' Weak positive link - slight trend, but not strong.')
else:
    print(' Very weak/no link - hours studied alone may not predict score.')
```

```python
import numpy as np

# Define your data arrays first
hours = np.array([2, 5, 1, 8, 4, 7, 3])
math = np.array([65, 80, 50, 95, 70, 88, 60])
english = np.array([70, 75, 60, 85, 80, 90, 65])
science = np.array([55, 82, 48, 91, 68, 85, 58])
```

```python
import numpy as np 

# np.corrcoef() gives a correlation matrix
# We want: correlation between hours studied and English score

corr_matrix_english = np.corrcoef(hours, english)
correlation_english = corr_matrix_english[0, 1]  # top-right value in the 2x2 matrix

print(f'Correlation between Hours Studied & English Score: {correlation_english:.4f}')
print()

if correlation_english > 0.5:
    print(' Strong positive link - studying more = better english!')
elif correlation_english > 0.2:
    print(' Weak positive link - slight trend, but not strong.')
else:
    print(' Very weak/no link - hours studied alone may not predict score.')
```

```python
import numpy as np 

# np.corrcoef() gives a correlation matrix
# We want: correlation between hours studied and English score


corr_matrix_science = np.corrcoef(hours, science)
correlation_science = corr_matrix_science[0, 1]  # top-right value in the 2x2 matrix

print(f'Correlation between Hours Studied & Science Score: {correlation_science:.4f}')
print()

if correlation_science > 0.5:
    print(' Strong positive link - studying more = better science!')
elif correlation_science > 0.2:
    print(' Weak positive link - slight trend, but not strong.')
else:
    print(' Very weak/no link - hours studied alone may not predict score.')
```

```python
# Add calculated columns back to the dataframe
df['Total_Score']   = total_score
df['Average_Score'] = np.round(student_avg, 2)
df['Result']        = np.where(df['Average_Score'] >= 50, 'Pass', 'Fail')

# Assign letter grades
def assign_grade(avg):
    if avg >= 90: return 'A+'
    elif avg >= 80: return 'A'
    elif avg >= 70: return 'B'
    elif avg >= 60: return 'C'
    elif avg >= 50: return 'D'
    else: return 'F'

df['Grade'] = [assign_grade(a) for a in df['Average_Score']]

# Save to a new CSV
df.to_csv('student_results.csv', index=False)

print(' Final results saved to student_results.csv!')
print()
print(' Grade Distribution:')
print(df['Grade'].value_counts().sort_index())
print()
print('Last 5 rows of final data:')
print(df.tail())
```
