```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Kathmandu-Weather-AQI-Dataset-2020.csv')
```

```python
# Step 2: Fix the Date format
# Computer needs to know 'Date and Time' is a real date, not just text or string. So we convert it to a datetime format.
df['Date and Time'] = pd.to_datetime(df['Date and Time'])

# Create a 'Month' column so we can group data by month
df['Month'] = df['Date and Time'].dt.month

# Step 3: Simple Analysis
# Let's calculate the average temperature for each month
monthly_temp = df.groupby('Month')['temperature_2m (°C)'].mean()

# Let's calculate total rain for each month
monthly_rain = df.groupby('Month')['rain (mm)'].sum()

print("Average Monthly Temperatures:")
print(monthly_temp)

# Step 4: Visualizing (Making Charts)

# Chart 1: Temperature over the year
plt.figure(figsize=(10, 5))
plt.plot(monthly_temp.index, monthly_temp.values, marker='o', color='red')
plt.title('Average Temperature in Kathmandu (Monthly)')
plt.xlabel('Month (1=Jan, 12=Dec)')
plt.ylabel('Temp (°C)')
plt.grid(True)
plt.show()

# Chart 2: Rainfall over the year
plt.figure(figsize=(10, 5))
plt.bar(monthly_temp.index, monthly_rain.values, color='blue')
plt.title('Total Rainfall in Kathmandu (Monthly)')
plt.xlabel('Month')
plt.ylabel('Rain (mm)')
plt.show()
```

```python
df['Month'] = df['Date and Time'].dt.month
df['Hour'] = df['Date and Time'].dt.hour

# --- ANALYSIS 1: Cloud Cover ---
# Average cloudiness per month
monthly_clouds = df.groupby('Month')['cloud_cover (%)'].mean()

# --- ANALYSIS 2: Humidity Morning vs. Afternoon ---
# Let's look at 8:00 AM (Morning) and 2:00 PM (Afternoon)
morning_hum = df[df['Hour'] == 8].groupby('Month')['relative_humidity_2m (%)'].mean()
afternoon_hum = df[df['Hour'] == 14].groupby('Month')['relative_humidity_2m (%)'].mean()

# --- ANALYSIS 3: Wind vs. Wind Gusts ---
# Normal wind vs. sudden strong gusts
monthly_wind = df.groupby('Month')['wind_speed_10m (km/h)'].mean()
monthly_gusts = df.groupby('Month')['wind_gusts_10m (km/h)'].mean()

# --- VISUALIZING ---

# Plot 1: Cloud Cover (Bar Chart)
plt.figure(figsize=(10, 4))
plt.bar(monthly_clouds.index, monthly_clouds.values, color='gray', alpha=0.6)
plt.title('How Cloudy is Kathmandu? (Monthly Average)')
plt.ylabel('Cloud Cover %')
plt.show()

# Plot 2: Morning vs Afternoon Humidity (Line Chart)
plt.figure(figsize=(10, 4))
plt.plot(morning_hum.index, morning_hum.values, label='8 AM (Morning)', marker='s')
plt.plot(afternoon_hum.index, afternoon_hum.values, label='2 PM (Afternoon)', marker='o')
plt.title('Humidity Comparison: Morning vs Afternoon')
plt.legend()
plt.ylabel('Humidity %')
plt.show()

# Plot 3: Wind Speed vs Gusts (Area Chart)
plt.figure(figsize=(10, 4))
plt.fill_between(monthly_wind.index, monthly_wind.values, monthly_gusts.values, color='orange', alpha=0.3, label='Gust Strength')
plt.plot(monthly_wind.index, monthly_wind.values, color='red', label='Average Wind Speed')
plt.title('Wind Speed vs. Sudden Gusts')
plt.legend()
plt.show()
```
